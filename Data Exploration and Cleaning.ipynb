**Choco Crunch Analytics — Data Exploration and Cleaning**

**Load the Raw Dataset**

The raw dataset collected from the Open Food Facts API is loaded from `chococrunch_raw.csv` into a Pandas DataFrame.

The display option is configured to show all columns when inspecting the dataset. The first few records are displayed to get an initial view of the data.

```python
import pandas as pd
pd.set_option('display.max_columns', None)  # Show all columns
df = pd.read_csv("chococrunch_raw.csv")

df.head()
```

```python
## Inspect the Dataset Columns

## The column names are examined to understand the variables available in the raw dataset and to identify the fields that will be used during the exploration and cleaning process.

df.columns
```

```python
## Check Dataset Dimensions

## The shape of the DataFrame is checked to determine the number of rows and columns currently present in the dataset.

df.shape
```

```python
## Check for Duplicate Records

## Duplicate records are identified by checking the DataFrame for completely duplicated rows.

## The total number of duplicate records is calculated to determine whether duplicate entries need to be removed from the dataset.

df.duplicated().sum()
```

```python
## Remove Duplicate Records

## The duplicate records identified in the previous step are removed from the DataFrame.

## The `inplace=True` parameter applies the change directly to `df` without creating a separate DataFrame.

df.drop_duplicates(inplace=True)
```

```python
## Verify the Updated Dataset Dimensions

## After removing duplicate records, the shape of the DataFrame is checked again to verify the updated number of rows and columns.

df.shape
```

```python
## Inspect Data Types and Dataset Structure

## The `info()` method is used to inspect the overall structure of the DataFrame.

## This provides information about the number of entries, column names, non-null values, and data types of each column.

df.info()
```

```python
## Generate Descriptive Statistics

## Descriptive statistics are generated for the numerical columns in the dataset.

## This provides an initial understanding of the distribution of numerical variables, including their count, mean, standard deviation, minimum, maximum, and quartile values.

df.describe()
```

```python
## Identify Missing Values

## The number of missing values in each column is calculated and sorted in descending order.

## This helps identify which columns contain the highest number of missing values and require further attention during the data-cleaning process.

df.isnull().sum().sort_values(ascending=False)
```

```python
## Calculate Missing Value Percentage

## The percentage of missing values in each column is calculated based on the total number of records in the dataset.

## Examining the missing values as percentages provides a clearer understanding of the extent of missing data across different columns.


(df.isnull().sum() / len(df)) * 100
```

**Check Missing Values in Product Name**

Since `product_name` is an important field for product-level analysis, we first check whether it contains any missing values.

```python
df['product_name'].isnull().sum()
```

**Check Missing Values in Brands**

The `brands` column is also important for brand-level analysis. We check this column separately to identify the number of missing values.

```python
df['brands'].isnull().sum()
```

**Identify Records with Missing Product Name and Brand**

Rows where both `product_name` and `brands` are missing do not contain either of the two key identification fields.

These records are checked separately so they can be removed before handling individual missing values.

```python
df[['product_name', 'brands']].isnull().all(axis=1).sum()
```

**Remove Records with Missing Product Name and Brand**

Records where both `product_name` and `brands` are missing are removed because neither product nor brand information is available for those records.

Rows where only one of the two fields is missing are retained because the available product or brand information can still be useful for analysis.

```python
df.dropna(subset=['product_name', 'brands'], how='all', inplace=True)
```

**Fill Remaining Missing Values**

After removing records where both fields were missing, the remaining missing values in `product_name` and `brands` are filled with `"Missing"`.

This preserves records where one field is available while clearly indicating that the other information was not provided in the original data.

```python
df[['product_name', 'brands']] = df[['product_name', 'brands']].fillna('Missing')
```

**Verify Product Name and Brand Cleaning**

The `product_name` and `brands` columns are displayed after cleaning to verify that the missing values have been handled correctly.

```python
df[['product_name', 'brands']].head(20)
```
