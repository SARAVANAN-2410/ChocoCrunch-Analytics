**Choco Crunch Analytics — Data Collection**

**Data Collection**

In this step, product data is collected from the Open Food Facts API.

The API is queried page by page for products in the **chocolates** category. Each request retrieves up to 100 records, and the process continues until approximately 12,000 records are collected.

The required product fields are selected directly from the API response, and the collected records are stored in `all_products` for further processing.

```python
import requests
from IPython.display import display

all_products = []
page = 1
page_size = 100
target_records = 12000

while len(all_products) < target_records:
    url = f"https://world.openfoodfacts.org/api/v2/search?categories=chocolates&fields=code,product_name,brands,nutriments&page_size={page_size}&page={page}"
    try:
        response = requests.get(url)
        response.raise_for_status()  # Raise an exception for bad status codes
        data = response.json()

        if not data.get('products'):
            print(f"No products found on page {page}. Stopping.")
            break # No more products

        all_products.extend(data['products'])
        print(f"Collected {len(all_products)} records so far.")
        page += 1

    except requests.exceptions.RequestException as e:
        print(f"Error fetching page {page}: {e}")
        break # Stop if there's an error


print(f"Finished collecting {len(all_products)} records.")
display(all_products)
```
