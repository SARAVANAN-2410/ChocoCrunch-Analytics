**Select Relevant Nutritional Features**

A list of relevant nutritional and product-quality features is defined for further processing.

These features include energy, carbohydrates, sugars, fats, proteins, fiber, salt, sodium, NOVA classification, nutrition score, and the estimated percentage of fruits, vegetables, and nuts.

Selecting the relevant columns helps create a focused dataset for the subsequent analysis.

```python
main_columns = [
    'nutriments.energy-kcal_value',
    'nutriments.energy-kj_value',
    'nutriments.carbohydrates_value',
    'nutriments.sugars_value',
    'nutriments.fat_value',
    'nutriments.saturated-fat_value',
    'nutriments.proteins_value',
    'nutriments.fiber_value',
    'nutriments.salt_value',
    'nutriments.sodium_value',
    'nutriments.nova-group',
    'nutriments.nutrition-score-fr',
    'nutriments.fruits-vegetables-nuts-estimate-from-ingredients_100g',
]
```

```python
## Create the Main Feature DataFrame

## The selected nutritional features are extracted from the cleaned DataFrame and copied into a new DataFrame named `df_main`.

## Creating a separate DataFrame allows the feature engineering process to be performed while keeping the original cleaned DataFrame available.

df_main = df[main_columns].copy()
```

```python
## Handle Missing Numerical Values

## Missing values in the selected nutritional features are filled using the median of each respective column.

## The median is used because it is less sensitive to extreme values and outliers compared with the mean, making it suitable for nutritional variables that may have skewed distributions.

for col in main_columns:
    median_val = df_main[col].median()  # median is robust to outliers
    df_main[col] = df_main[col].fillna(median_val)
```

```python
## Verify Missing Values After Imputation

## The missing-value percentage is calculated again after median imputation to verify the effect of the missing-value treatment.

(df_main.isnull().sum() / len(df)) * 100
```

**Create Sugar-to-Carbohydrate Ratio**

A new feature, `sugar_to_carb_ratio`, is created by calculating the ratio of sugar content to total carbohydrate content.

This derived feature provides an additional measure of how much of the total carbohydrate content is contributed by sugars.

The resulting missing values are replaced with `0`, which also safely handles cases where the carbohydrate value is zero and the division produces an undefined result.

```python
# Calculate ratio of sugar to total carbohydrates
df['nutriments.sugar_to_carb_ratio'] = df['nutriments.sugars_value'] / df['nutriments.carbohydrates_value']

# Handle divide-by-zero safely (if carbohydrates = 0)
df['nutriments.sugar_to_carb_ratio'] = df['nutriments.sugar_to_carb_ratio'].fillna(0)
```

**Create Calorie Categories**

A categorical feature named `calorie_category` is created from the energy value in kilocalories.

Products are grouped into three categories:

- **Low** — Less than 100 kcal
- **Moderate** — Between 100 and 200 kcal
- **High** — Greater than 200 kcal

This converts the continuous calorie value into an easy-to-interpret categorical feature.

```python
def calorie_category(kcal):
    if kcal < 100:
        return 'Low'
    elif 100 <= kcal <= 200:
        return 'Moderate'
    else:
        return 'High'

df['nutriments.calorie_category'] = df['nutriments.energy-kcal_value'].apply(calorie_category)
```

**Create Sugar Categories**

A categorical feature named `sugar_category` is created based on the sugar content of each product.

The products are grouped into three categories:

- **Low Sugar** — Less than 5
- **Moderate Sugar** — Between 5 and 15
- **High Sugar** — Greater than 15

This feature makes it easier to categorize and compare products based on their sugar content.

```python
def sugar_category(sugar):
    if sugar < 5:
        return 'Low Sugar'
    elif 5 <= sugar <= 15:
        return 'Moderate Sugar'
    else:
        return 'High Sugar'

df['nutriments.sugar_category'] = df['nutriments.sugars_value'].apply(sugar_category)
```

**Identify Ultra-Processed Products**

A binary categorical feature named `is_ultra_processed` is created using the NOVA food classification.

Products with a NOVA group value of `4` are classified as **Yes**, indicating that they are ultra-processed. Other values are classified as **No**.

This feature can be used to analyze the relationship between ultra-processing and the nutritional characteristics of chocolate products.

```python
df['nutriments.is_ultra_processed'] = df['nutriments.nova-group'].apply(lambda x: 'Yes' if x == 4 else 'No')
```

**Combine Product and Derived Features**

Additional product-identification and engineered features are added to `df_main`.

These include:

- Product code
- Product name
- Brand
- Sugar-to-carbohydrate ratio
- Calorie category
- Sugar category
- Ultra-processed classification

This combines the selected nutritional variables with the product information and newly created features into a single analysis-ready DataFrame.

```python
# Columns to move from df to df_main
additional_columns = [
    'code',           # product_code
    'product_name',   # product_name
    'brands',         # brand
    'nutriments.sugar_to_carb_ratio',
    'nutriments.calorie_category',
    'nutriments.sugar_category',
    'nutriments.is_ultra_processed'    
]

for col in additional_columns:
    df_main[col] = df[col]

# Verify
df_main.head()
```
